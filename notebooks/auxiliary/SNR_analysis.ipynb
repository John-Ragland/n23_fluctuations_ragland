```python
dss = tools.open_receptions()
```

```python
noise_slices = []
for k in range(23):
    noise_slices.append(dss[k].sel({'time':slice(90,100)}).std('time').values)
```

```python
snrs = 20*np.log10(np.array(noise_slices).flatten())
```

```python
finite_mask = np.isfinite(snrs)
```

```python
_ = plt.hist(snrs[finite_mask].flatten(), bins=40, density=True)
plt.xlabel('Noise Power Estimate')
```
