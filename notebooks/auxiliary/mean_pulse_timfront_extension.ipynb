# Mean Pulse

```python
from nescafe23_fluctuations import tools
from matplotlib import pyplot as plt
import numpy as np
import xarray as xr
from dotenv import load_dotenv, find_dotenv
import os
import cmps
import colorsys

_ = load_dotenv(find_dotenv())

%config InlineBackend.figure_format = 'retina'
```

```python
fn = f"{os.environ['NESCAFE23_DATADIR']}dataproducts/mean_pulse_lower.nc"
mean_pulse_lower = xr.open_dataarray(fn)

fn = f"{os.environ['NESCAFE23_DATADIR']}dataproducts/mean_pulse_upper.nc"
mean_pulse_upper = xr.open_dataarray(fn)
```

```python
fig = plt.figure(figsize=(10,3))

(20*np.log10(np.abs(mean_pulse_lower))).plot(x='time', vmax=120, vmin=95, cmap=cmps.timefront(), cbar_kwargs={'label':'dB rel [$\\mu$Pa]'}, rasterized=True)
(20*np.log10(np.abs(mean_pulse_upper))).plot(x='time', vmax=120, vmin=95, cmap=cmps.timefront(), add_colorbar=False, rasterized=True)

plt.xlim([100,102])
plt.ylim([1200,100])
plt.xlabel('depth [m]')
plt.ylabel('time [s]')

fig.savefig('../auxiliary/figures/mean_pulse.pdf', dpi=300)
```
