# Looking at dz/d$\theta$

```python
import pickle
from dotenv import load_dotenv, find_dotenv
import os
import numpy as np
import matplotlib.pyplot as plt
import pygenray as pr

load_dotenv(find_dotenv())
```

```python
fn = f"{os.environ['NESCAFE23_DATADIR']}/dataproducts/ray_run_forward_back.pkl"

with open(fn, 'rb') as f:
    ray_runs = pickle.load(f)

rays = ray_runs['rays']
eigs = ray_runs['eigs']
```

```python
plt.scatter(rays.ts[:,-1], rays.zs[:,-1], s=10, lw=0)
plt.plot(rays.ts[:,-1], rays.zs[:,-1], lw=1, c='k')

plt.xlim([102.2, 103])
plt.ylim([-1500,0])
```

```python
# compute central difference for every eigen ray

eig_launch_angles = eigs.launch_angles[0]
```

```python
rays_lower = pr.shoot_rays(2300, 0, eig_launch_angles - 1e-7, 153e3, 1000, ray_runs['envi'])
rays_upper = pr.shoot_rays(2300, 0, eig_launch_angles + 1e-7, 153e3, 1000, ray_runs['envi'])
```

```python
dtheta = 2e-7
dz = rays_upper.zs - rays_lower.zs

dzdtheta = np.abs(dz/dtheta)
```

```python
plt.plot(dzdtheta[15,:])
```

```python
eigs.ps[0].shape
```

```python
P1 = np.sqrt((1/153e3) * (np.abs(eigs.ps[0][:,0]) / np.abs(eigs.ps[0][:,-1])) * (1/np.abs(dzdtheta[:,-1])))
```

```python
20*np.log10(P1/1e-6)
```
