```python

plt.plot(rvs[ray_id][0]['rs']/1000, rvs[ray_id][0]['zs'])

plt.fill_between(rvs[ray_id][0]['rs']/1000, rvs[ray_id][0]['zs']-np.sqrt(rvs[ray_id][0]['Rf2']), rvs[ray_id][0]['zs']+np.sqrt(rvs[ray_id][0]['Rf2']), alpha=0.5)
```

```python
fig = plt.figure(figsize=(10,10))

plot_variables = ['zs', 'Cray', 'Nray', 'N0B_ray', 'mu2', 'wronk',]
for k, var in enumerate(plot_variables):
    plt.subplot(len(plot_variables),2,k+1)

    plt.plot(rvs[ray_id][0]['rs']/1000, rvs[ray_id][0][var])
    plt.title(var)

plt.tight_layout(pad=0)
```
