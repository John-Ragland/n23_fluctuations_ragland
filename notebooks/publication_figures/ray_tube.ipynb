# Ray Seperation Figure

```python
import nescafe23 as n23
from n23_fluctuations_ragland import tools
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

%config InlineBackend.figure_format = 'retina'
```

```python
ray_vars = tools.compute_n23_ray_variables()
```

```python
ray_ids = ['-8', '+7', '-7', '-6', '+6']
integrand = {}
delta_zr = np.array([200])

ray_r = {}
ray_z = {}
zeta1 = {}

for ray_id in ray_ids:
    integrand[ray_id] = []
    ray_r[ray_id] = []
    ray_z[ray_id] = []
    zeta1[ray_id] = []

    for rv in ray_vars[ray_id]: 
        integrand[ray_id].append(tools.compute_phase_structure_integrand(rv, delta_zr))
        ray_r[ray_id].append(rv['rs'])
        ray_z[ray_id].append(rv['zs'])
        zeta1[ray_id].append(rv['RT_zeta_1'])
```

```python
sns.set_palette("colorblind")

plt.rcParams.update({'font.size':8})
plt.rcParams["text.usetex"] = True

fig, axes = plt.subplots(3, 1, figsize=(6.67, 4))


plt.sca(axes[2])

for ray_id in ['-7', '+6', '-6']:

    if ray_id == '-6':
        plt.plot(ray_vars[ray_id][0]['rs']/1000, integrand[ray_id][0][:,-1], label=f'${ray_id}_{{i}}$', alpha=0.85)
        plt.plot(ray_vars[ray_id][1]['rs']/1000, integrand[ray_id][1][:,-1], label=f'${ray_id}_{{ii}}$', alpha=0.85)
        #plt.plot(ray_vars[ray_id][2]['rs']/1000, integrand[ray_id][2][:,-1], label=f'${ray_id}_{{iii}}$', alpha=0.85)
    else:
        plt.plot(ray_vars[ray_id][0]['rs']/1000, integrand[ray_id][0][:,-1], label=ray_id, alpha=0.85)

plt.title('')



plt.xlabel('range [km]')
plt.title('$D(\Delta z)$ integrand', loc='left')
plt.ylabel('')

plt.sca(axes[0])

for ray_id in ['-7', '+6', '-6']:

    if ray_id == '-6':
        plt.plot(ray_vars[ray_id][0]['rs']/1000, -ray_z[ray_id][0], label=f'${ray_id}_{{i}}$', alpha=0.85)
        plt.plot(ray_vars[ray_id][1]['rs']/1000, -ray_z[ray_id][1], label=f'${ray_id}_{{ii}}$', alpha=0.85)

    else:
        plt.plot(ray_vars[ray_id][0]['rs']/1000, -ray_z[ray_id][0], label=ray_id, alpha=0.85)

plt.ylim([3000,000])
plt.ylabel('depth [m]')

leg = plt.legend(ncols=4, framealpha=1)

# change the line width for the legend
for line in leg.get_lines():
    line.set_linewidth(4.0)

plt.title('Ray paths', loc='left')
axes[0].set_xticklabels([])
plt.sca(axes[1])


for ray_id in ['-7', '+6', '-6']:

    if ray_id == '-6':
        plt.plot(ray_vars[ray_id][0]['rs']/1000, zeta1[ray_id][0], label=f'${ray_id}_{{i}}$', alpha=0.85)
        plt.plot(ray_vars[ray_id][1]['rs']/1000, zeta1[ray_id][1], label=f'${ray_id}_{{ii}}$', alpha=0.85)

    else:
        plt.plot(ray_vars[ray_id][0]['rs']/1000, zeta1[ray_id][0], label=ray_id, alpha=0.85)

plt.ylabel('ray separation [m]')
plt.title('Ray separation $\\xi_1$', loc='left')
axes[1].set_xticklabels([])

plt.tight_layout(pad=0)
fig.savefig('../../figures/phase_structure_integrand.pdf', bbox_inches='tight')
```

```python
from scipy.signal import argrelextrema
import numpy as np
```

```python
ray_ids = ['-7', '-6', '+6']


plt.rcParams.update({'font.size':8})
plt.rcParams["text.usetex"] = True

fig = plt.figure(figsize=(3.37,3))




for ray_id in ray_ids:

    rs = ray_vars[ray_id][0]['rs']
    zs = ray_vars[ray_id][0]['zs']

    # Find local maxima in rs
    rs_max_idx = argrelextrema(np.gradient(np.array(zs)), np.greater)[0]
    
    # Extract the coordinates for scatter plot
    # Using "either" — adjust to "both_idx" if you prefer
    scatter_rs = np.array(rs)[rs_max_idx]
    ridxs = np.argmin(np.abs(rs[:,None] - scatter_rs[None,:]), axis=0)
    scatter_zeta1 = ray_vars[ray_id][0]['RT_zeta_1'][ridxs]

    plt.plot(ray_vars[ray_id][0]['rs']/1000, ray_vars[ray_id][0]['RT_zeta_1'], label=ray_id)

    plt.scatter(scatter_rs/1000, scatter_zeta1)
plt.legend(loc='lower left', ncols=3)
```
